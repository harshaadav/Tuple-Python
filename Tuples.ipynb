```python
thistuple = ("apple", "banana", "cherry")
thistuple = "apple", "banana", "cherry" #Tuples can also be created without the parentheses
print(thistuple)
```

```python
thistuple = ("apple", "banana", "cherry", "apple", "cherry")#allows duplicates
print(thistuple)
```

```python
thistuple = ("apple", "banana", "cherry")
print(len(thistuple))#length of tuple
```

```python
mytuple = ("apple", "banana", "cherry")
print(type(mytuple))#to find data type
```

```python
thistuple = tuple(("apple", "banana", "cherry")) # note the double round-brackets
print(thistuple)#Using the tuple() method to make a tuple
```

```python
thistuple = ("apple", "banana", "cherry")
print(thistuple[1])#Print the second item in the tuple
print(thistuple[-1])#Negative indexing means start from the end
```

```python
thistuple = ("apple", "banana", "cherry", "orange", "kiwi", "melon", "mango")
print(thistuple[2:5])#displaying a specific range of tuple
```

```python
thistuple = ("apple", "banana", "cherry", "orange", "kiwi", "melon", "mango")
print(thistuple[:4])#returns the items from the beginning to 4 ,not including 4
print(thistuple[2:])#returns the items from 2, including 2, to the end
print(thistuple[-4:-1])
```

```python
x = ("apple", "banana", "cherry")
y = list(x)#Convert the tuple into a list to be able to change it since it is immutable 
y[1] = "kiwi"
x = tuple(y)
print(x)
```

```python
thistuple = ("apple", "banana", "cherry")
y = ("orange",)
thistuple += y#allowed to add tuples to tuples
print(thistuple)
```

```python
thistuple = ("apple", "banana", "cherry")
del thistuple#The del keyword can delete the tuple completely
```

```python
thistuple = ("apple", "banana", "cherry","apple")
print(thistuple.count("apple"))#Returns the number of times a specified value occurs in a tuple
```

```python
thistuple = ("apple", "banana", "cherry", "orange", "kiwi", "melon", "mango")
print(thistuple.index("orange"))#Searches the tuple for a specified value and returns the position of where it was found
```
